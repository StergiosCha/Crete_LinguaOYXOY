## 9. Εξερεύνηση αποτελεσμάτων

```python
import json

if result_files:
    with open(result_files[-1]) as f:
        data = json.load(f)

    print(f"Σύνολο: {len(data['results'])} αποτελέσματα")
    print(f"Μοντέλα: {sorted(set(r['model'] for r in data['results']))}")
    print(f"Φαινόμενα: {sorted(set(r['phenomenon'] for r in data['results']))}")

    # Quick table
    print(f"\n{'Phenomenon':<25s} {'Condition':<30s} {'Model':<20s} {'Mean':>6s} {'SD':>6s}")
    print('-' * 90)
    for r in sorted(data['results'], key=lambda x: (x['phenomenon'], x['condition'], x['model'])):
        m = f"{r['mean']:.1f}" if r['mean'] else '---'
        s = f"{r['sd']:.1f}" if r.get('sd') else '---'
        print(f"{r['phenomenon']:<25s} {r['condition']:<30s} {r['model']:<20s} {m:>6s} {s:>6s}")
```

## 10. Γράφημα

```python
try:
    import matplotlib.pyplot as plt
    import numpy as np
except ImportError:
    !pip install matplotlib
    import matplotlib.pyplot as plt
    import numpy as np

if result_files:
    with open(result_files[-1]) as f:
        data = json.load(f)

    # Group by phenomenon
    phenomena = sorted(set(r['phenomenon'] for r in data['results']))

    for phenom in phenomena:
        items = [r for r in data['results'] if r['phenomenon'] == phenom and r['mean'] is not None]
        if not items:
            continue

        conditions = sorted(set(r['condition'] for r in items))
        models = sorted(set(r['model'] for r in items))

        fig, ax = plt.subplots(figsize=(max(10, len(conditions) * 1.2), 5))
        x = np.arange(len(conditions))
        width = 0.8 / len(models)

        for i, model in enumerate(models):
            means = []
            for cond in conditions:
                match = [r for r in items if r['condition'] == cond and r['model'] == model]
                means.append(match[0]['mean'] if match else 0)
            ax.bar(x + i * width, means, width, label=model)

        ax.set_ylabel('Mean Rating')
        ax.set_title(phenom.upper())
        ax.set_xticks(x + width * (len(models) - 1) / 2)
        ax.set_xticklabels(conditions, rotation=45, ha='right', fontsize=8)
        ax.legend(fontsize=8)
        ax.set_ylim(0, 7.5)
        ax.axhline(y=4, color='gray', linestyle='--', alpha=0.5)
        plt.tight_layout()
        plt.show()
```
